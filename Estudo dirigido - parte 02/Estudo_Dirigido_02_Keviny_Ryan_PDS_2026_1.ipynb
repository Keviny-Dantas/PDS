1. Fundamentação Teórica de Sistemas LTI

Em Processamento Digital de Sinais, a análise de sistemas é simplificada drasticamente quando tratamos de sistemas Lineares e Invariantes no Tempo (LTI). A importância dessa classe de sistemas reside no fato de que eles são completamente caracterizados por sua resposta a uma entrada fundamental: o impulso unitário.

1.1 Resposta ao Impulso $h[n]$

A resposta ao impulso, denotada por $h[n]$, corresponde à saída do sistema quando a entrada é o impulso discreto $\delta[n]$. Matematicamente, se o sistema é representado pelo operador $T$, então:

$$h[n] = T\{\delta[n]\}$$

A potência conceitual desse parâmetro advém do fato de que qualquer sequência de entrada $x[n]$ pode ser decomposta como uma soma ponderada de impulsos deslocados no tempo:

$$x[n] = \sum_{k=-\infty}^{+\infty} x[k] \delta[n-k]$$

Devido à propriedade da linearidade (superposição) e da invariância no tempo, se conhecemos a resposta do sistema a um único impulso, conhecemos sua resposta a qualquer combinação deles. Portanto, conhecer $h[n]$ equivale a conhecer completamente a dinâmica do sistema LTI.

2. A Operação de Convolução Discreta

A convolução é a operação matemática que permite determinar a saída $y[n]$ de um sistema LTI para qualquer entrada $x[n]$ conhecida. Ela representa a superposição das respostas do sistema a cada amostra da entrada, ponderadas pela amplitude de cada amostra.

2.1 Formulação Matemática

A saída $y[n]$ é obtida pela convolução entre a entrada e a resposta ao impulso:

$$y[n] = x[n] * h[n]$$

De forma expandida, a soma de convolução é expressa como:

$$y[n] = \sum_{k=-\infty}^{+\infty} x[k] h[n - k]$$

Nesta expressão, para cada instante $n$, realizamos uma soma ponderada onde a sequência $h[k]$ é invertida no tempo e deslocada. Esse resultado mostra que a saída pode ser interpretada como uma composição ponderada de versões deslocadas da resposta ao impulso.

2.2 Aplicações Práticas em Engenharia
Em sistemas reais, como os baseados em microcontroladores e processamento de sensores, a convolução aparece em:

Filtragem de ruídos: Remoção de interferências de alta frequência em sinais de sensores.

Suavização de medições: Redução de flutuações espúrias em dados amostrados.

Média móvel: Um tipo de convolução onde $h[n]$ é uma janela retangular, usada para extrair a tendência de um sinal.

Análise da dinâmica: Modelagem de como sistemas físicos (motores, circuitos) reagem a estímulos discretos.

3. Equações de Diferenças Lineares

Muitos sistemas discretos, especialmente filtros digitais implementados em hardware (como RTL em SystemVerilog ou C para STM32), são descritos por equações de diferenças. Elas descrevem a evolução temporal da saída de forma recursiva.Um exemplo clássico de sistema de primeira ordem é:

$$y[n] - ay[n-1] = bx[n]$$

Ou, isolando a saída atual:

$$y[n] = ay[n-1] + bx[n]$$

Essas equações são análogas às equações diferenciais no caso contínuo. Elas mostram que a saída atual $y[n]$ depende não apenas da entrada atual $x[n]$, mas também do estado anterior do sistema $y[n-1]$. Isso caracteriza sistemas com memória.

4. Resposta Transitória e Memória do Sistema

A resposta transitória corresponde ao comportamento inicial do sistema imediatamente após a aplicação de uma entrada ou mudança de estado.

Em sistemas com memória (como filtros IIR), a saída leva um intervalo de tempo (amostras) para atingir um regime estável ou permanente.

O tempo de duração do transitório está diretamente ligado aos polos do sistema e à duração da sua resposta ao impulso.

5. Causalidade e Estabilidade BIBO

Para a implementação prática em engenharia, duas propriedades são fundamentais:

5.1 Causalidade

Um sistema é causal se a saída em um instante $n$ depende apenas de valores presentes ($n$) e passados ($n-1, n-2, \dots$) da entrada. Em termos de resposta ao impulso, um sistema LTI é causal se:

$$h[n] = 0, \quad \forall n < 0$$

5.2 Estabilidade BIBOUm sistema é estável no sentido BIBO (Bounded-Input, Bounded-Output) se toda entrada limitada em amplitude produzir uma saída também limitada. Para sistemas LTI, a condição necessária e suficiente para estabilidade BIBO é que a resposta ao impulso seja absolutamente somável:

$$\sum_{n=-\infty}^{+\infty} |h[n]| < \infty$$

Se essa soma convergir para um valor finito, o sistema é estável. Caso contrário, o sistema pode apresentar oscilações crescentes ou saturação de hardware, o que é crítico em projetos de controle e sistemas embarcados.

### Atividade 1: Interpretação Conceitual

**1) O que significa afirmar que um sistema é linear e invariante no tempo?**
Um sistema LTI (Linear e Invariante no Tempo) obedece ao princípio da superposição (a resposta a uma soma de entradas ponderadas é a soma das respostas individuais ponderadas) e a propriedade de invariância temporal (um atraso na entrada causa um atraso idêntico na saída, sem alterar o formato da resposta).

**2) Por que a resposta ao impulso é suficiente para caracterizar um sistema LTI?**
Porque qualquer sinal discreto pode ser modelado como uma soma de impulsos escalonados e atrasados. Devido à linearidade e invariância no tempo, basta conhecer a resposta do sistema a um único impulso ($h[n]$) para calcular a resposta a qualquer entrada através da operação de convolução.

**3) Qual o significado físico da convolução em sistemas discretos?**
Fisicamente, a convolução é o processo de "filtragem". Ela determina o valor da saída em um instante específico acumulando (somando) as memórias passadas e presentes do sistema, onde o peso de cada amostra passada é ditado pelo formato da resposta ao impulso.

**4) Qual a diferença entre resposta transitória e regime permanente?**
A **resposta transitória** é a reação de acomodação inicial do sistema após a aplicação de um sinal. Em sistemas estáveis, ela decai a zero com o tempo. O **regime permanente** é o estado de equilíbrio que se estabelece após o desaparecimento do transitório.

**5) O que se entende por sistema causal?**
É um sistema onde a saída no instante atual $n$ depende estritamente de entradas presentes e passadas ($k \le n$), sem antecipar o futuro. Matematicamente, exige que $h[n] = 0$ para $n < 0$.

**6) O que se entende por sistema estável?**
Um sistema estável no sentido BIBO (*Bounded-Input, Bounded-Output*) garante que qualquer entrada finita resultará em uma saída finita. Para isso, a resposta ao impulso deve ser absolutamente somável ($\sum |h[n]| < \infty$).

### Atividade 2: Cálculo Manual de Convolução

Dadas as sequências:
* $x[n] = \{1, 2, 1\}$
* $h[n] = \{1, 1\}$

A soma de convolução é $y[n] = \sum_{k=-\infty}^{+\infty} x[k]h[n-k]$. Considerando $n=0$ na primeira amostra:
* $y[0] = x[0]h[0] = 1 \times 1 = 1$
* $y[1] = x[0]h[1] + x[1]h[0] = (1 \times 1) + (2 \times 1) = 3$
* $y[2] = x[1]h[1] + x[2]h[0] = (2 \times 1) + (1 \times 1) = 3$
* $y[3] = x[2]h[1] = 1 \times 1 = 1$

**Sequência Resultante:** $y[n] = \{1, 3, 3, 1\}$.
**Significado:** A operação combinou amostras vizinhas da entrada, atuando como um filtro passa-baixas rudimentar que alargou (suavizou) o pulso original no domínio do tempo.

### Atividade 3: Sistema Descrito por Equação de Diferenças

Dado o sistema recursivo: $y[n] = 0.8y[n-1] + x[n]$ com repouso inicial ($y[-1] = 0$).
Para encontrar os 6 primeiros valores de $h[n]$, aplicamos a entrada $x[n] = \delta[n]$:

* $h[0] = 0.8(0) + 1 = 1$
* $h[1] = 0.8(1) + 0 = 0.8$
* $h[2] = 0.8(0.8) + 0 = 0.64$
* $h[3] = 0.8(0.64) + 0 = 0.512$
* $h[4] = 0.8(0.512) + 0 = 0.4096$
* $h[5] = 0.8(0.4096) + 0 = 0.32768$

**Análise:** * **Estabilidade:** A resposta ao impulso tem o formato $h[n] = (0.8)^n u[n]$. Sendo a base menor que 1, a série geométrica converge para um valor finito, logo o sistema é **estável**.
* **Causalidade:** A saída atual só depende da entrada atual e da saída imediatamente anterior, sem valores futuros. O sistema é **causal**.

## Atividade 4: Implementação Computacional e Interpretação

**Parte 2: Análise de Regime Permanente**

```python
%%octave
clc;
clear;
close all;
x = [1 2 1];
h = [1 1];
y = conv(x,h);
disp('Sequencia x[n]:');
disp(x);
disp('Sequencia h[n]:');
disp(h);
disp('Convolucao y[n] = x[n] * h[n]:');
disp(y);
n = 0:length(y)-1;
stem(n,y,'filled');
grid on;
xlabel('n');
ylabel('y[n]');
title('Resultado da convolucao discreta');
```

## Atividade 5: Suavização de Sinais (Filtro de Média Móvel)

Nesta etapa, utilizamos a convolução para processar um sinal ruidoso através de um filtro de média móvel de 3 pontos. Abaixo, apresentamos a fundamentação teórica e as respostas aos questionamentos do roteiro:

**1) Comparação dos resultados:**
Ao comparar o sinal original $x[n]$ com o sinal resultante $y[n]$, observa-se que os picos abruptos presentes na entrada (como o salto de 2 para 5 na segunda amostra) foram atenuados. O sinal de saída apresenta uma curva mais contínua e suave, demonstrando o efeito de redução de variância entre amostras adjacentes.

**2) Por que este filtro atua como um suavizador?**
Este filtro atua como um suavizador porque, para cada instante $n$, a saída $y[n]$ é calculada como a média aritmética da amostra atual e das suas vizinhas imediatas. Como flutuações rápidas (ruído) tendem a variar de forma brusca para cima e para baixo, a soma ponderada de amostras sucessivas faz com que essas componentes de alta frequência se cancelem mutuamente, preservando apenas as variações mais lentas (baixas frequências) do sinal.

**3) Interpretação da Resposta ao Impulso $h[n]$:**
A resposta ao impulso utilizada, $h[n] = \{1/3, 1/3, 1/3\}$, define uma janela retangular no domínio do tempo. Na prática da Engenharia, este é o filtro FIR mais simples possível para eliminar ruídos térmicos ou interferências de leitura em sensores analógicos, sendo fundamental em projetos de instrumentação e monitoramento.

Código base:

```python
%%octave
clc;
clear;
close all;
x = [2 5 4 6 8 7 5 4];
h = (1/3)*[1 1 1];
y = conv(x,h);
figure;
stem(0:length(x)-1,x,'filled');
grid on;
xlabel('n');
ylabel('x[n]');
title('Sinal original');
figure;
stem(0:length(y)-1,y,'filled');
grid on;
xlabel('n');
ylabel('y[n]');
title('Sinal suavizado por convolucao');
```

```python
%%octave
% ==========================================
% ATIVIDADE 5: Implementacao Aprimorada (Subplots)
% Filtro de Media Movel
% ==========================================
clc; clear; close all;

% 1. Definicao do sinal original (simulando aquisicao com flutuacoes)
x = [2 5 4 6 8 7 5 4];

% 2. Definicao do filtro de media movel (janela de ponderacao de 3 amostras)
h = (1/3) * [1 1 1];

% 3. Aplicacao da convolucao para filtragem
y = conv(x, h);

% Exibicao dos vetores no console
disp('Sinal de entrada x[n]:'); disp(x);
disp('Sinal suavizado y[n]:'); disp(y);

% 4. Plotagem Comparativa
figure('Position', [100, 100, 800, 600]);

% Grafico 1 (Superior): Sinal Original
subplot(2, 1, 1);
stem(0:length(x)-1, x, 'filled', 'r', 'LineWidth', 1.5, 'MarkerSize', 6);
grid on;
title('Sinal Original x[n] (Aquisicao com Flutuacoes)', 'FontSize', 12);
ylabel('Amplitude', 'FontSize', 11);

% Grafico 2 (Inferior): Sinal Suavizado
subplot(2, 1, 2);
stem(0:length(y)-1, y, 'filled', 'b', 'LineWidth', 1.5, 'MarkerSize', 6);
grid on;
title('Sinal Suavizado y[n] (Apos Filtro de Media Movel)', 'FontSize', 12);
xlabel('Tempo discreto n (Amostras)', 'FontSize', 11);
ylabel('Amplitude', 'FontSize', 11);
```

**Por que atua como suavizador?**
O filtro atua como suavizador porque flutuações rápidas (ruídos de alta frequência) tendem a oscilar para cima e para baixo. Ao calcular a média na janela de amostras, essas variações abruptas se cancelam, mitigando os picos agudos e mantendo apenas a tendência principal do sinal.

### Atividade 6: Análise de Causalidade e Estabilidade

Nesta etapa, analisamos as propriedades fundamentais de quatro sistemas LTI distintos.

**a) $y[n] = x[n] + x[n-1]$**
* **Causalidade:** Causal. A saída $y[n]$ depende estritamente do instante atual $n$ e do instante passado $n-1$.
* **Estabilidade:** Estável. A resposta ao impulso é $h[n] = \delta[n] + \delta[n-1]$. A soma dos módulos é finita ($1 + 1 = 2 < \infty$).

**b) $y[n] = x[n+1]$**
* **Causalidade:** Não Causal. O sistema antecipa a entrada, pois necessita da amostra futura $n+1$ para calcular a saída atual.
* **Estabilidade:** Estável. A resposta ao impulso é um impulso deslocado $h[n] = \delta[n+1]$, cuja soma dos módulos é finita ($1 < \infty$).

**c) $h[n] = (0.5)^n u[n]$**
* **Causalidade:** Causal. A presença da função degrau unitário $u[n]$ garante que a resposta ao impulso seja nula para tempos negativos ($h[n] = 0$ para $n < 0$).
* **Estabilidade:** Estável. Trata-se de uma série geométrica com razão $|0.5| < 1$. A soma infinita converge para um valor finito ($S = 2$).

**d) $h[n] = 2^n u[-n]$**
* **Causalidade:** Não Causal. A função $u[-n]$ reflete o degrau, gerando valores não nulos em todo o intervalo negativo do tempo ($n \le 0$).
* **Estabilidade:** Estável. Apesar de não causal, a soma em tempo reverso decai à medida que $n \to -\infty$, equivalendo à soma de $(0.5)^n$ de $0$ a $\infty$, o que resulta em convergência finita.

## Desafio Proposto: Convolução em Sistemas de Aquisição de Dados

Na prototipagem de monitores de sinais vitais voltados para cuidados pós-internação, a aquisição de dados de oximetria e frequência cardíaca utilizando sensores acoplados a microcontroladores (como a integração serial de um sensor BerryMed BM1000C, por exemplo) frequentemente sofre com a introdução de ruídos de leitura de alta frequência.

A convolução do sinal bruto com a resposta ao impulso de um **filtro de média móvel** aprimora a confiabilidade da medição da seguinte maneira:

1. **O Papel de $h[n]$:** A resposta ao impulso do filtro de média define o tamanho da "janela de observação" e distribui pesos iguais (ex: $1/M$) para cada amostra da janela. Isso instrui o algoritmo do sistema embarcado a compor a saída baseada em um bloco recente de dados.
2. **Mecanismo de Suavização:** Ruídos oriundos de variações de contato do sensor ou flutuações térmicas tendem a orbitar em torno do valor real, com média tendendo a zero em intervalos curtos. A convolução calcula a média local contínua, forçando o cancelamento dos picos positivos e negativos adjacentes, preservando a variação lenta (que representa os verdadeiros sinais fisiológicos do paciente).
3. **Limitações e Trade-offs:** A principal limitação deste filtro em ambientes críticos é o atraso (delay) introduzido na saída. Se o paciente apresentar uma queda aguda real e repentina de oxigenação, o filtro aplicará um "borramento" nessa rampa de descida. Isso significa que o microcontrolador precisará de mais milissegundos ou segundos processando os quadros seguintes para registrar a mudança, o que deve ser calibrado cuidadosamente para não comprometer o acionamento de alertas.
